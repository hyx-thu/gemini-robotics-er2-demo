### Dynamixel Control

#### dxl_manager Examples

````python
# This function drives as fast as possible to the position listed. This Example does not define a speed. For an example that does see Emotes.
# Test of DXL Manager class

import dxl_manager as DXL
from time import sleep

# set up a list with all of DXL IDs in it
MY_DXL = [1,2]

# instantiate the dxl_manager
bothead = DXL.DXLManager(MY_DXL, DEVICENAME='/dev/ttyUSB1') # may need to change according to the Device

# read curr pos and enable torque on all motors in your chain. IF torque is not enabled it will not move.
for i in MY_DXL:
    bothead.read_curr_pos(i)
    bothead.enable_torque(i)

# send goal positions to the head.
print('Writing goal pose')
bothead.write_goal_pose(1,180)
bothead.write_goal_pose(2,180)
sleep(0.5)

# Step through some poisitons:
# for d in range(180):
#    bothead.readClear()
#    bothead.read_curr_pos(1)
#    bothead.write_goal_pose(1,d)
#    sleep(0.05)


# bothead.disable_torque(1)
# bothead.disable_torque(2)
````

````python
# Close disable torque and close the port
bothead.disable_torque(1)
bothead.disable_torque(2)
bothead.portHandler.closePort()
````

````python
# This is an example using the write_goal_pose_with_speed() function.
import dxl_manager as DXL
from time import sleep

# set up a list with all of DXL IDs in it
MY_DXL_CHAIN = [1,2]

# instantiate the dxl_manager
bothead = DXL.DXLManager(MY_DXL=MY_DXL_CHAIN, DEVICENAME='/dev/ttyUSB0')
# enable torque
for i in MY_DXL_CHAIN:
  bothead.enable_torque(i)

# read current postions
bothead.read_all_curr_pos(MY_DXL_CHAIN)
# send to desired positions
bothead.write_goal_pose_with_speed(1,180,150)
bothead.write_goal_pose_with_speed(2,180,150)
sleep(0.3)
# after waiting a small amount of time read position to verify that you drove to the intended positions
bothead.read_all_curr_pos(MY_DXL_CHAIN)
````

### Using the TinyBot class

````python
# An example using the TinyBot class. This class uses the lower level dxl_manager class to crate some little emotes and other useful 'canned' actions.

from tinybot import TinyBot as HRI
from time import sleep

MY_DXL_CHAIN = [1, 2]
DEVICENAME = '/dev/ttyUSB0'

# Initialize
print("Initializing TinyBot...")
head_controller = HRI(DEVICENAME=DEVICENAME, DXL_CHAIN=MY_DXL_CHAIN,BAUDRATE = 115200)


print('Moving to home position')
head_controller.home_pose()
sleep(0.5)

print('Nodding')
head_controller.nod(speed=180)
sleep(0.5)

print('Shaking head no')
head_controller.no(speed=200)
````
